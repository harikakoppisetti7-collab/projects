```python
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)

IQR = Q3 - Q1

outliers = ((df.select_dtypes(include="number") < (Q1 - 1.5 * IQR)) |
            (df.select_dtypes(include="number") > (Q3 + 1.5 * IQR)))

outliers.sum()
```
